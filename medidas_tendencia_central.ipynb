# Medidas de Tendência Central
As medidas de tendência central são métricas usadas para verificar os comportamentos dos dados em relação ao eixo horizontal (x) do gráfico de frequência.

Elas buscam descrever o comportamento dos dados a partir da sua frequência e apresentar valores representativos do conjunto de dados analisado.

**Principais Medidas:**
> 1. Média Populacional
> 1. Média Amostral
> 1. Mediana
> 1. Moda

### Qual a importância de realizar os testes?
é essencial porque elas ajudam a entender o comportamento geral dos dados.

**Para ter uma visão Geral dos dados**
> As medidas de tendência central indicam um valor típico dos dados, ajudando a entender qual é o comportamento médio ou comum da variável analisada. Exemplo: Se estamos analisando salários de funcionários, a média pode indicar um valor típico pago pela empresa.

**Para detectar assimetria e outliers**
> Se a média e a mediana forem muito diferentes, pode indicar a presença de valores extremos (outliers) ou que a distribuição dos dados é assimétrica. Exemplo: Em um conjunto de salários onde a maioria ganha R$ 3.000, mas um diretor ganha R$ 100.000, a média pode ser elevada, enquanto a mediana refletirá melhor o valor típico.

**Para escolher a melhor medida de resumo**
> Em distribuições simétricas, a média representa bem os dados.
> Se os dados forem assimétricos ou contiverem outliers, a mediana pode ser um melhor indicador da tendência central.
> Se há valores repetidos, a moda pode ser útil, por exemplo, para identificar o tamanho de roupa ou sapato mais vendido em uma loja.

### Importante
O principal campo da estatística responsável pelos métodos de coleta e descriçã de dados é a **Amostragem** e a **Estatística Descritiva**.


## Média Populacional

μ, média populacional.
Xn, variável estudada.

## Mediana
É o valor central de um conjunto de dados ordenado. Se houver um número ímpar de elementos, a mediana é o valor central. Se houver um número par, a mediana é a média dos dois valores centrais.

Exemplo (n ímpar): Para os dados 10, 15, 20, 25, 30, a mediana é 20.
Exemplo (n par): Para os dados 10, 15, 20, 25, a mediana é (15+20)/2 = 17,5


### Preparação

```python
# Bibliotecas
import pandas as pd
from scipy.stats import skew
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Carregando Dados
brasileiro = pd.read_csv("dados_demograficos_brasileiros.csv",  delimiter=",", encoding="utf-8")
print(brasileiro.head())
```

```python
# Preparando os Dados
brasileiro["Sexo"] = brasileiro["Sexo"].replace({0: "Masculino", 1: "Feminino"})
print(brasileiro.head(5))
```

### Aplicação

#### Mediana

```python
# Comparação entre a mediana e a Média (IDADE)
median_brasileiro_idade = brasileiro["Idade"].median()
med_brasileiro_idade = brasileiro["Idade"].mean()
print(f"Mediana da Idade: {median_brasileiro_idade: .2f}",f"Média da Idade: {med_brasileiro_idade: .2f}")
```

```python
# Comparação entre a mediana e a média (ANOS DE ESTUDO)
median_brasileiro_anosestudos = brasileiro["Anos de Estudo"].median()
med_brasileiro_anosestudos = brasileiro["Anos de Estudo"].mean()
print(f"Mediana dos Anos de Estudo: {median_brasileiro_anosestudos: .2f} e Média dos Anos de Estudo: {med_brasileiro_anosestudos: .2f}")
```

##### Por que comparar a Mediana com a Média?
Como pode ser observado acima, o cálculo da Mediana e da Média retornou valores muito próximos. Seria certo dizer que a distribuição dos dados é simétrica?

**Resposta:** A proximidade entre a média e a mediana pode ser um indicativo de que a distribuição dos dados é aproximadamente simétrica, mas não garante isso com certeza.

🧐 Por que isso acontece?
> **Distribuições simétricas** (ex.: normal) geralmente têm média ≈ mediana.
> **Distribuições assimétricas** (ex.: enviesadas à direita ou esquerda) tendem a ter maior diferença entre média e mediana.

##### O que verificar para confirmar a simetria?
**Cálculo do Coeficiente de Assimetria (Skewness)**
> Skewness (ou assimetria) é uma medida estatística que descreve o grau de distorção de uma distribuição em relação a uma distribuição simétrica (normal). Ela indica se os valores dos dados estão mais concentrados em uma das extremidades da distribuição.

1. Se skewness = 0, a distribuição é perfeitamente simétrica.
1. Se 0 < skewness < 0.5, a distribuição é aproximadamente simétrica.
1. Se skewness > 0.5, há uma assimetria significativa para a direita.
1. Se skewness < -0.5, há uma assimetria significativa para a esquerda.

```python
# Assimetria de Anos de Estudo
skewness = skew(brasileiro["Anos de Estudo"])

# Os dados apresentam distribuição simétrica?
if skewness == 0:
    print(f"Distribuição perfeitamente simétrica: {skewness: .2f}")
elif 0 < skewness < 0.5:
    print(f"A distribuição é ligeiramente simétrica: {skewness: .2f}")
elif skewness > 0.5:
    print(f"Há uma assimetria significativa para a direita: {skewness: .2f}")
elif skewness < 0.5:
    print(f"Há uma assimetria significativa para a esquerda: {skewness: .2f}")
```

```python
# Histograma
sns.histplot(brasileiro['Anos de Estudo'], bins=10, kde=True)
plt.axvline(brasileiro['Anos de Estudo'].median(), color='red', linestyle='dashed', label='Mediana')
plt.axvline(brasileiro['Anos de Estudo'].mean(), color='blue', linestyle='dashed', label='Média')
plt.legend()
plt.show()
```

```python
# Assimetria da Idade
skewness = skew(brasileiro['Idade'])

# Os dados apresentam distribuição simétrica?
if skewness == 0:
    print(f"Distribuição perfeitamente simétrica: {skewness:.2f}")
elif 0 < skewness < 0.5:
    print (f"Distribuição é aproximadamente simétrica: {skewness:.2f}")
elif skewness > 0.5:
    print(f"Há uma assimetria significativa para a direita: {skewness:.2f}")
elif skewness < -0.5:
    print(f"Há uma assimetria significativa para a esquerda: {skewness:.2f}")
```

```python
# Histograma
sns.histplot(brasileiro['Idade'], bins=10, kde=True)
plt.axvline(brasileiro['Idade'].median(), color='red', linestyle='dashed', label='Mediana')
plt.axvline(brasileiro['Idade'].mean(), color='blue', linestyle='dashed', label='Média')
plt.legend()
plt.show()
```

## Moda
É o valor que aparece com maior frequência no conjunto de dados.

Exemplo: Para os dados 3, 5, 7, 7, 9, a moda é 7, pois ocorre duas vezes.
Se nenhum valor se repete, dizemos que não há moda.
Se houver mais de um valor com a maior frequência, a distribuição pode ser bimodal (duas modas) ou multimodal.

### Preparação

```python
# Bibliotecas
import pandas as pd
```

```python
# Carregamento da Base
vendas = pd.read_csv('vgsales.csv', delimiter=',', encoding='utf-8')
print(vendas.head(5))
```

### Aplicação

```python
# Calcular a moda (gênero mais frequente)
moda_genre = vendas['Genre'].mode()[0]

# Contar quantas vezes a moda (gênero mais frequente) aparece
count_moda = (vendas['Genre'] == moda_genre).sum()

print(f"O gênero mais frequente é: {moda_genre}, que aparece {count_moda} vezes.")
```

```python
# Calcular a moda (Publisher mais frequente)
moda_publisher = vendas["Publisher"].mode()[0]

# Contar quantas vezes a moda (publisher mais frequente) aparece
count_moda = (vendas['Publisher'] == moda_publisher).sum()

print(f"O Publisher mais frequente é: {moda_publisher}, que aparece {count_moda} vezes.")
```
